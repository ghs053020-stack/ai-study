# Ollama 채팅 화면 — 노트북 안에 띄우기

셀을 **위에서부터 차례로** 실행하면 마지막 셀 아래에 채팅 화면이 뜹니다.
화면을 끄려면 마지막 셀의 `demo.close()` 를 실행하세요.

```python
import gradio as gr
from openai import OpenAI

openai = OpenAI(base_url="http://localhost:11434/v1", api_key="ollama")
MODEL = "llama3.2"
```

```python
def to_text(content):
    # Gradio 버전에 따라 content 가 문자열 또는 [{"type": "text", "text": ...}] 목록으로 온다
    if isinstance(content, str):
        return content
    return "".join(part.get("text", "") for part in content if isinstance(part, dict))

def chat(message, history):
    messages = [{"role": h["role"], "content": to_text(h["content"])} for h in history]
    messages.append({"role": "user", "content": message})
    response = openai.chat.completions.create(model=MODEL, messages=messages)
    return response.choices[0].message.content
```

```python
demo = gr.ChatInterface(fn=chat, title=f"Ollama 채팅 ({MODEL})")
demo.launch()
```

```python
demo.close()
```
